**1.- Grafique las siguientes funciones en un mismo plano **cartesiano**: f(x) = x^2  y  g(x)= 2x + 3**

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir el intervalo de valores de x
x = np.linspace(-2, 4, 400)

# 2. Definir las funciones
f = x**2
g = 2*x + 3

# 3. Crear el gráfico
plt.plot(x, f, label="f(x) = x²", color="blue")
plt.plot(x, g, label="g(x) = 2x + 3", color="red")

# 4. Agregar elementos al gráfico
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gráfico de dos funciones")
plt.grid(True)
plt.legend()

# 5. Mostrar el gráfico
plt.show()
```

**2.- Encuentre los valores de x para los cuales ambas funciones son iguales. Es decir presentan el mismo valor en el eje vertical. Vale decir: x^2 = 2x +3**

```python
import numpy as np
from scipy.optimize import fsolve

def funcion(x):
  return x**2 - (2*x + 3)

x = np.linspace(-2, 1000, 1)

sol = fsolve(funcion, x)

print(sol[0])
```

**5.- Encuentre los valores de x para los cuales x^2 = 16**

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir el intervalo de valores de x
x = np.linspace(-5, 5, 400)

# 2. Definir las funciones
f = x**2
g = np.full_like(x, 16) # en caso de una función constante

# 3. Crear el gráfico
plt.plot(x, f, label="f(x) = x²", color="blue")
plt.plot(x, g, label="g(x) = 16", color="green")

# 4. Agregar elementos al gráfico
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Gráfico de f(x) = x² y g(x) = 16")
plt.grid(True)
plt.legend()

# 5. Mostrar el gráfico
plt.show()
```

Según el grafico los valores son x= -4  y   x= 4

```python
import numpy as np
from scipy.optimize import fsolve

def funcion(x):
  return x**2 - (16)

x1 = np.linspace(-5, 1000, 1)
x2 = np.linspace(2, 1000, 1)

sol1 = fsolve(funcion, x1)
sol2 = fsolve(funcion, x2)

print(sol1[0])
print(sol2[0])
```
