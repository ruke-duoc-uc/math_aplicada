```python
def s2(n):
    return 2*n**2+3*n+5
for i in range(1,100):
    no_superar = 500
    actual = s2(i)
    if actual >= no_superar:
        print(f"Limite en {i-1}")
        print(s2(i-1))
        print(f"Siguiente{i}")
        print(s2(i))
        break
print(f"50 registros {s2(50)}")
```
