e) Aqui hay que resolver lo siguiente: 1650*x + 4200 = 103200

```python
import numpy as np
from scipy.optimize import fsolve

def funcion(x):
  return  1650*x +4200 - (103200)

x = np.linspace(0, 100, 1) # va buscando la solución desde x=0 hasta x=100, nótese que pueden existir varias soluciones según sea el caso

sol = fsolve(funcion, x)

print(" Si el cobro fue de $103.200, la cantidad de gigabyte transmitidos fue de: ", sol[0], "GB")
```

```python
# Algoritmo para graficar dos funciones: f(x) = 100 * np.exp(-0.1 * x) y g(x) = 2 * x + 3
import numpy as np
import matplotlib.pyplot as plt

# 1. Definir el intervalo de valores de x
x = np.linspace(0, 12, 400)

# 2. Definir las funciones
f = -10*x + 115


# 3. Crear el gráfico de las funciones
plt.plot(x, f, label="f = -10*x + 115", color="blue")


# 4. Agregar elementos al gráfico
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("tiempo (min)")
plt.ylabel("Cant de combustible ( litros)")
plt.title("Gráfico de C(x)")
plt.grid(True)
plt.legend()

# 5. Mostrar el gráfico
plt.show()
```

c) Graficar ambas funciones y ver dónde se intersectan

```python

import numpy as np
import matplotlib.pyplot as plt

# 1. Definir el intervalo de valores de x
x = np.linspace(0, 5, 400)

# 2. Definir las funciones
f = -2*x**2+16*x+5
g = 8 * x + 5

# 3. Crear el gráfico de las funciones
plt.plot(x, f, label="f(x) = -2*x**2+16*x+5", color="blue")
plt.plot(x, g, label="g(x) = 8 * x + 5", color="red")

# 4. Agregar elementos al gráfico
plt.axhline(0, color="black", linewidth=0.8)
plt.axvline(0, color="black", linewidth=0.8)

plt.xlabel("tiempo de carga ( horas)")
plt.ylabel("Cant de energía (kwh)")
plt.title("Gráfico de f(x) y g(x)")
plt.grid(True)
plt.legend()

# 5. Mostrar el gráfico
plt.show()
```

ambos vehiculos habrán obtenido la misma cantidad de energía a las 4 horas de carga ( x=4)
