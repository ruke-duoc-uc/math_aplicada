### Solucion
En este caso decidimos utilizar 3 graficos separados para mostrar el progreso, considerando las marcas de tiempo que nos piden

```python
import matplotlib.pyplot as plt
import numpy as np

# Variable base para la velocidad de transmisión (100 Mb/s)
velocidad_mbps = 100

# Eje X común: Unidades de tiempo de 0 a 60
x = np.linspace(0, 60, 100)

# Cálculos para cada escala
y_sec = velocidad_mbps * x           # 100 Mb/s
#y_min = (velocidad_mbps * 60) * x    # 6,000 Mb/min
#y_hr = (velocidad_mbps * 3600) * x   # 360,000 Mb/h

# Crear un único gráfico
plt.plot(x, y_sec, label="Segundos (100 Mb/s)", color="purple", linewidth=2)
#plt.plot(x, y_min, label="Minutos (6,000 Mb/min)", color="dodgerblue", linewidth=2)
#plt.plot(x, y_hr, label="Horas (360,000 Mb/h)", color="orange", linewidth=2)

plt.xlabel("Unidades de tiempo")
plt.ylabel("Megabits transmitidos (Mb)")
plt.title("Transmisión de datos según la unidad de tiempo (Base: 100 Mbps)")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

velocidad_mbps = 100
x = np.linspace(0, 60, 100)
y_min = (velocidad_mbps * 60) * x
plt.plot(x, y_min, label="Minutos (6,000 Mb/min)", color="dodgerblue", linewidth=2)
plt.xlim(1.4,1.6)
plt.ylim(8000,10000)

# Aqui se establece un plot aparte para destacar el punto
# de interes en cada caso
x_punto = 1.5
y_punto= (velocidad_mbps * 60) * x_punto
plt.plot(x_punto,y_punto,marker='.',color='red', markersize=8,label= "Minuto / Megabits (1.5)")    # 6,000 Mb/min

plt.xlabel("Unidades de tiempo")
plt.ylabel("Megabits transmitidos (Mb)")
plt.title("Transmisión de datos según la unidad de tiempo (Base: 100 Mbps)")

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

velocidad_mbps = 100
x = np.linspace(0, 60, 100)
y_min = (velocidad_mbps * 60) * x    # 6,000 Mb/min

plt.plot(x, y_min, label="Minutos (6,000 Mb/min)", color="dodgerblue", linewidth=2)

# Destacar el punto en el minuto 1.5
x_destacado = 1.5
y_destacado = (velocidad_mbps * 60) * x_destacado
plt.plot(x_destacado, y_destacado, marker='.', color='red', markersize=8)
plt.vlines(x_destacado, ymin=0, ymax=y_destacado, color='gray', linestyle='--')
plt.hlines(y_destacado, xmin=0, xmax=x_destacado, color='gray', linestyle='--')
plt.xlim(1.4, 1.6)
plt.ylim(8000, 10000)
plt.xlabel("Unidades de tiempo")
plt.ylabel("Megabits transmitidos (Mb)")
plt.title("Transmisión de datos según la unidad de tiempo (Base: 100 Mbps)")

plt.show()
```
