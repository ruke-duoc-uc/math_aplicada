1) Olvide considerar el contexto, lo importante es el alcance positivo, no nos interesa cuando "avanzan hacia atras"

```python
import numpy as np
import matplotlib.pyplot as plt

def s(x):
    return -1.6*x**2+1.3*x+141.3
def r(x):
    return 1.4*x**2-5.6*x+12.6

t = np.linspace(-10, 10, 200)
y = s(t)

t2 = np.linspace(-10, 10, 200)
y2 = r(t2)

plt.plot(t, y, color="blue", linewidth=2, label="f(t)")
plt.plot(t2, y2, color="blue", linewidth=2, label="f(t)")

plt.show()
```

Aqui tenemos la "distancia" entre las funciones, osea, el espacio entre ellas.
Este arreglo de numeros nos interesa para llegar a sus intersecciones, y en este caso, el punto donde ambos vehiculos van a la misma velocidad

roots soluciona la ecuacion cuadratica para que el resultado sea 0

- return -1.6*x**2+1.3*x+141.3
- return 1.4*x**2-5.6*x+12.6

sort le da un orden a los resultados (intersecciones), puede contenerse en una variable como lista, o valores individuales

```python
raices = np.roots([-3.0, 6.9, 128.7])
x_min, x_max = np.sort(raices)

print(f"s(x) > r(x) en el intervalo: ({x_max:.4f}, {x_min:.4f})")
```

4) Es importante marcar bien las multiplicaciones y elevaciones, el resultado cambia drasticamente

```python
import math

def d(t):
    return 50*math.e**0.3*t
print(f"Incorrecto: {d(6)}")
```

```python
import math

def d(t):
    return 50*(math.e**(0.3*t))
print(f"Correcto: {d(6)}")
```

```python
def s2(n):
    return 2*n**2+3*n+5
for i in range(1,100):
    no_superar = 500
    actual = s2(i)
    if actual >= no_superar:
        print(f"Limite en {i-1}")
        print(s2(i-1))
        print(f"Siguiente {i}")
        print(s2(i))
        break
print(f"50 registros {s2(50)}")
```
